# Dataset selection

I chose [Video Game Sales with Ratings](https://www.kaggle.com/datasets/rush4ratio/video-game-sales-with-ratings) for Project 1. Each row is a game released on one platform. The aim is to group releases by sales, regional audience and reviews.

The file is `Video_Games_Sales_as_at_22_Dec_2016.csv`, version 2. It contains 16,719 rows and 16 columns. Sales are in millions of units and describe a historical snapshot, not current sales.

## Features

There are 14 usable original features after excluding `Name`, an identifier, and `Global_Sales`, which repeats the regional sales total.

| Type | Features | Count |
| --- | --- | --- |
| Continuous | `NA_Sales`, `EU_Sales`, `JP_Sales`, `Other_Sales` | 4 |
| Scores | `Critic_Score`, `User_Score` | 2 |
| Discrete numerical | `Year_of_Release`, `Critic_Count`, `User_Count` | 3 |
| Categorical | `Platform`, `Genre`, `Publisher`, `Developer`, `Rating` | 5 |

This meets the selection minimums of 2,000 rows, 10 features, 3 continuous features and 2 categorical features.

The main issue is missing reviews. After treating `tbd` as missing, 54.6% of user scores and 51.3% of critic scores are unavailable. Sales are strongly skewed, and four release years are later than 2016. These issues are examined in [the analysis notebook](analysis.ipynb), which also downloads the data and verifies the counts.
