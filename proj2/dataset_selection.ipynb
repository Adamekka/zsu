# Dataset selection

I chose [Spotify Tracks Dataset](https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset) for Project 2. The aim is to predict a song's recorded Spotify popularity from its audio features and genre. The target is `popularity`, a score from 0 to 100. This is a regression task.

The file is `dataset.csv`, version 1. It contains 114,000 rows and 21 columns, including an exported row index. There are 114 genres, including drum-and-bass, metal, electronic and anime.

## Features

There are 15 usable original features after excluding `popularity`, the row index, `track_id` and the names `artists`, `album_name` and `track_name`. There are no exact duplicate columns.

| Type | Features | Count |
| --- | --- | --- |
| Continuous numerical | `danceability`, `energy`, `loudness`, `speechiness`, `acousticness`, `instrumentalness`, `liveness`, `valence`, `tempo` | 9 |
| Other numerical | `duration_ms` | 1 |
| Categorical | `explicit`, `key`, `mode`, `time_signature`, `track_genre` | 5 |

For example, `tempo` is BPM, `loudness` is in dB, `instrumentalness` estimates the absence of vocals and `valence` describes how positive the music sounds. `mode` means major or minor.

This meets the minimums of 2,000 rows, 10 features, 3 continuous features and 2 categorical features.

Expected preprocessing:

- Remove the exported row index and the 450 duplicate rows found after ignoring it.
- Encode categorical inputs. Musical key and time signature are categories despite being stored as numbers.
- Check zero duration and tempo values and numerical outliers before modeling.
- Keep repeated `track_id` values in the same train/test split. Fit encoding and any numerical scaling on training data only.
